## Código de análisis

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq

def analizar_ensayo(ruta_csv):
    # 1. Cargar datos ignorando las lineas de comentarios de metadatos
    datos = np.loadtxt(ruta_csv, delimiter=',', comments='#')
    tiempo = datos[:, 0]
    amplitud = datos[:, 1]

    # 2. Calcular parametros criticos del muestreo digital
    n_muestras = len(tiempo)
    dt = tiempo[1] - tiempo[0] # Paso de tiempo delta_t
    fs = 1.0 / dt # Frecuencia de muestreo calculada (Hz)

    # 3. Computar la Transformada Rapida de Fourier (FFT)
    espectro_crudo = fft(amplitud)
    magnitud = np.abs(espectro_crudo) / n_muestras # Normalizacion de amplitud

    # 4. Generar el eje de frecuencias correcto de Fourier
    frecuencias = fftfreq(n_muestras, d=dt)

    # 5. Filtrar solo la mitad positiva del espectro (Simetria de Fourier)
    mitad = n_muestras // 2
    frecuencias_pos = frecuencias[:mitad]
    magnitud_pos = 2.0 * magnitud[:mitad] # Multiplicacion por 2 por perdida de energia mitad negativa

    # 6. Graficacion Automatizada de Resultados
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8,6))

    # Grafica en el Dominio del Tiempo
    ax1.plot(tiempo, amplitud, color='darkblue', linewidth=1.5)
    ax1.set_title("Analisis Temporal (Datos Crudos del CSV)")
    ax1.set_xlabel("Tiempo (segundos)")
    ax1.set_ylabel("Amplitud")
    ax1.grid(True, linestyle='--')

    # Grafica en el Dominio de Fourier
    ax2.plot(frecuencias_pos, magnitud_pos, color='crimson', linewidth=2)
    ax2.set_title("Analisis Espectral (Espacio de Fourier)")
    ax2.set_xlabel("Frecuencia (Hz)")
    ax2.set_ylabel("Magnitud Espectral")
    ax2.set_xlim(0, 600) # Limite visual enfocado en el rango util del instrumento
    ax2.grid(True, linestyle='--')

    plt.tight_layout()
    plt.show()
```
